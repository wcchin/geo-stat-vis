## 1. Generate and inspect the observations

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KernelDensity
rng = np.random.default_rng(47)
points = np.vstack([rng.normal((250, 300), 65, (100, 2)),
                    rng.normal((730, 690), 100, (160, 2))])
points = np.clip(points, 0, 1000)
coords = np.linspace(0, 1000, 55)
xx, yy = np.meshgrid(coords, coords)
grid = np.column_stack((xx.ravel(), yy.ravel()))
print("Observations:", len(points))
```

## 2. Analyse and evaluate

KDE smooths observed events into an estimated intensity surface. The evaluation grid is not the set of observations.

**TODO 1 — bandwidth.** Fit a Gaussian KDE with `bandwidth=60`, evaluate it at `grid`, and display the resulting surface.

```python
kde = KernelDensity(kernel="gaussian", bandwidth=60).fit(points)
surface = np.exp(kde.score_samples(grid)).reshape(xx.shape)
fig, ax = plt.subplots()
im = ax.imshow(surface, origin="lower", extent=(0,1000,0,1000))
ax.scatter(points[:,0], points[:,1], c="white", s=3, alpha=.35)
ax.set(title="KDE: bandwidth 60 m", xlabel="x (m)", ylabel="y (m)")
plt.colorbar(im, ax=ax, label="Estimated probability density")
plt.show()
```

**TODO 2 — sensitivity.** Repeat with bandwidth 160. Compare the number and apparent breadth of concentrated regions. Why is the smoothest surface not automatically the best?

```python
wide = KernelDensity(kernel="gaussian", bandwidth=160).fit(points)
wide_surface = np.exp(wide.score_samples(grid)).reshape(xx.shape)
fig, ax = plt.subplots()
im = ax.imshow(wide_surface, origin="lower", extent=(0,1000,0,1000))
ax.set(title="KDE: bandwidth 160 m", xlabel="x (m)", ylabel="y (m)")
plt.colorbar(im, ax=ax, label="Estimated probability density")
plt.show()
```

## 3. Worked interpretation

These surfaces are probability density estimates (integrating to approximately 1 over the full plane), not event counts per square kilometre. Their scales and interpretation must be stated explicitly; bandwidth can merge locally distinct concentrations.
