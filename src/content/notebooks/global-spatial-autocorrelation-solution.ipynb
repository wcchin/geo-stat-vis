## 1. Set up observations and spatial weights

Rows of the rook-contiguity weight matrix are standardised to sum to one. The units form a fictional regular grid, not real administrative areas.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(5230)
n_side = 6
n = n_side**2
coords = np.array([(x,y) for y in range(n_side) for x in range(n_side)])
# Synthetic neighbouring-area pattern: a smooth geographic gradient plus noise.
values = 15 + 1.8*coords[:,0] + 1.2*coords[:,1] + rng.normal(0, 1, n)
W = np.zeros((n,n))
for i, (x,y) in enumerate(coords):
    for j, (xx,yy) in enumerate(coords):
        if abs(x-xx)+abs(y-yy)==1: W[i,j]=1
W = W/W.sum(axis=1,keepdims=True)
z = values-values.mean()
S0=W.sum()
observed = (n/S0)*(z@W@z)/(z@z)
print("Global Moran I:",round(observed,3))
```

**TODO 1:** Calculate the row-standardised spatial lag of centred values and draw a Moran scatterplot.

```python
lag=W@z
fig,ax=plt.subplots()
ax.scatter(z,lag)
ax.axhline(0,color="grey",lw=.7);ax.axvline(0,color="grey",lw=.7)
ax.set(xlabel="Deviation from mean",ylabel="Neighbour-average deviation",title="Moran scatterplot")
plt.show()
```

**TODO 2:** Generate 999 permutation statistics by shuffling the centred values while keeping W fixed. Report the upper-tail Monte Carlo p-value.

```python
permutations=np.empty(999)
for k in range(999):
    zz=rng.permutation(z)
    permutations[k]=(n/S0)*(zz@W@zz)/(zz@zz)
p_upper=(1+np.count_nonzero(permutations>=observed))/(len(permutations)+1)
print("Upper-tail permutation p-value:",p_upper)
```

## Worked interpretation

The permutation distribution is conditional on the chosen values and weights. Global Moran's I does not explain the mechanism producing spatial similarity; a different neighbour definition may change the result.
