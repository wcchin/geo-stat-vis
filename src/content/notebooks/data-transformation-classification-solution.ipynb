# Transforming and classifying HDB resale prices

**Question:** How do transformations and classification methods affect our representation of the same values?

**Concepts:** Min-max normalisation, z-scores, quantile and equal-interval classes. **Data:** 2023 HDB teaching extract.

Try the tasks before opening the companion solution. Class membership is sensitive to the scheme and should not be confused with intrinsic groups in the data.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Works in a repository checkout and remotely (including Colab).
name = "hdb_data_2023.csv.xz"
candidates = [
    Path("../datasets/singapore_hdb_ura") / name,
    Path("src/content/datasets/singapore_hdb_ura") / name,
    Path("datasets/singapore_hdb_ura") / name,
]
source = next((p for p in candidates if p.is_file()), "https://raw.githubusercontent.com/wcchin/geo-stat-vis/8cf3ddb0a7eff7d94018940e91f32d7482f32aaa/src/content/datasets/singapore_hdb_ura/" + name)
df = pd.read_csv(source, index_col=0)
assert {"resale_price", "resale_age"}.issubset(df.columns)
print(f"Loaded {len(df):,} records from {source}")
df[["resale_price", "resale_age"]].describe()
```

## 1. Reference summaries

Start by checking price range and distribution.

```python
price = df["resale_price"].dropna().astype(float)
print(price.describe())
print("Missing values:", df["resale_price"].isna().sum())
```

## 2. Exercise — normalisation

**TODO:** Compute (a) min-max normalisation to [0, 1] and (b) standard scores. Verify the minimum/maximum of the first and the mean/standard deviation of the second.

```python
price_minmax = (price - price.min()) / (price.max() - price.min())
price_z = (price - price.mean()) / price.std()
print("Min-max range:", price_minmax.min(), price_minmax.max())
print("Z mean/SD:", price_z.mean(), price_z.std())
```

## 3. Exercise — classification

**TODO:** Classify resale prices into 5 classes using both equal-width intervals and quantiles. Compare counts per class. Why are these not the same?

```python
equal_width = pd.cut(price, bins=5, include_lowest=True)
quantiles = pd.qcut(price, q=5, duplicates="drop")
print("Equal-width counts:\n", equal_width.value_counts().sort_index())
print("Quantile counts:\n", quantiles.value_counts().sort_index())
```

## Interpretation

Explain one circumstance in which a quantile map could mislead readers about differences in actual prices. You do not need geometry to answer this representation question.

## Worked interpretation

Quantiles tend to contain similar numbers of observations but can group very different numerical spans. Equal intervals retain comparable value ranges but may have uneven counts.
