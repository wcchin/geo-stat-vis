## 1. Generate and inspect the observations

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, DBSCAN
rng = np.random.default_rng(123)
points = np.vstack([rng.normal((0, 0), .35, (90, 2)),
                    rng.normal((3, 0), .4, (110, 2)),
                    rng.normal((1.4, 2.9), .4, (90, 2)),
                    rng.uniform((-1.5, -1.5), (4.5, 4), size=(25, 2))])
fig, ax = plt.subplots()
ax.scatter(points[:, 0], points[:, 1], s=10)
ax.set(title="Synthetic spatial observations", xlabel="x", ylabel="y")
plt.show()
```

## 2. Analyse and evaluate

The generating locations are known to us, but are not provided to the models. Cluster labels are arbitrary identifiers; they do not name geographical categories.

**TODO 1 — K-means.** Partition points with `n_clusters=3`. Plot labelled groups and note how outliers are handled.

```python
km = KMeans(n_clusters=3, random_state=0, n_init=10).fit(points)
fig, ax = plt.subplots()
ax.scatter(points[:,0], points[:,1], c=km.labels_, s=10)
ax.scatter(km.cluster_centers_[:,0], km.cluster_centers_[:,1], marker="x", c="black")
ax.set(title="K-means with k=3")
plt.show()
```

**TODO 2 — DBSCAN.** Fit `DBSCAN(eps=0.3, min_samples=6)`. Count the distinct non-noise clusters and noise observations. Try a larger eps and explain changes.

```python
db = DBSCAN(eps=0.3, min_samples=6).fit(points)
labels = db.labels_
n_clusters = len(set(labels) - {-1})
n_noise = int((labels == -1).sum())
print("DBSCAN clusters:", n_clusters, "noise points:", n_noise)
fig, ax = plt.subplots()
ax.scatter(points[:,0], points[:,1], c=labels, s=12)
ax.set(title="DBSCAN (noise labelled -1)")
plt.show()
```

## 3. Worked interpretation

K-means assigns every observation to one of k groups, while DBSCAN may label sparse observations as noise. Spatial clusters depend on the coordinates, distance metric, scaling and parameters; neither algorithm explains why the pattern exists.
