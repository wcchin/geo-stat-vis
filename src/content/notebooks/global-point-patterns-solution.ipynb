## 1. Generate and inspect the observations

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import cKDTree
rng = np.random.default_rng(5230)
window = (0., 1000., 0., 1000.)  # x min/max, y min/max in fictional metres
observed = np.vstack([rng.normal((300, 300), 60, size=(75, 2)),
                      rng.normal((720, 700), 65, size=(75, 2))])
observed = np.clip(observed, 0, 1000)
fig, ax = plt.subplots()
ax.scatter(observed[:, 0], observed[:, 1], s=9)
ax.set(xlim=(0, 1000), ylim=(0, 1000), aspect="equal", title="Observed synthetic events")
plt.show()
print("Points:", len(observed))
```

## 2. Analyse and evaluate

First, estimate where most observations occur. Then ask whether the distribution is consistent with uniform CSR within the fixed window.

**TODO 1 — quadrats.** Use a 5×5 grid with `np.histogram2d`. Report the range of quadrat counts; explain why a few high counts alone cannot establish a statistically significant departure from CSR.

```python
counts, _, _ = np.histogram2d(observed[:, 0], observed[:, 1], bins=5, range=[[0, 1000], [0, 1000]])
print("Quadrat range:", counts.min(), counts.max())
fig, ax = plt.subplots()
ax.imshow(counts.T, origin="lower", extent=[0,1000,0,1000], interpolation="nearest")
ax.set(title="Quadrat event counts", xlabel="x (m)", ylabel="y (m)")
plt.show()
```

**TODO 2 — NNA.** Find each point's nearest *other* neighbour and compute Clark–Evans R using the uncorrected infinite-plane CSR expectation. Explain the edge-effect limitation.

```python
nearest = cKDTree(observed).query(observed, k=2)[0][:, 1]
intensity = len(observed) / 1_000_000
r_expected = 0.5 / np.sqrt(intensity)
R = nearest.mean() / r_expected
print(f"Mean NND: {nearest.mean():.2f} m; uncorrected R: {R:.3f}")
```

## 3. Worked interpretation

An R below 1 is descriptive evidence of short distances relative to the infinite-plane expectation, not by itself a valid significance test. Boundary effects and the chosen observation window matter.
