Use a recent Python 3 environment in JupyterLab, Jupyter Notebook or Google Colab. Google Colab is optional: the public notebooks should not require access to the GE5230 course Google Drive.

## 2. Check your environment


```python
import sys
import importlib.util
print('Python:', sys.version.split()[0])
for package in ('numpy', 'pandas', 'matplotlib', 'scipy', 'geopandas', 'sklearn', 'libpysal', 'esda'):
    available = importlib.util.find_spec(package) is not None
    print(f'{package}: {"installed" if available else "not installed"}')
```
